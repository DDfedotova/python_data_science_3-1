В табличке представленной выше можем сравнить, количество подтвержденных заболеваемостей, количество выздоровевших и количество умерших в Африке, Европе и тд.

```python
corrmat = data[['Deaths / 100 Cases', 'Recovered / 100 Cases', 'Confirmed']].corr()
corrmat
```

Построив матрицу парных корреляций Пирсона, можно сказать о том, что связи между показателями смертности к числу случаев болезни, выздоравления к число случаев болезни и подтверждение заболеваемости очень слабые. Это говорит о том, взаимосвязь меду данными показателями почти не прослеживается и рассматривать корреляцию в дальнейшем не имеет смысла. 

```python
corrmat = data[['Deaths / 100 Cases', 'Recovered / 100 Cases', 'Deaths / 100 Recovered']].corr()
corrmat
```

Если рассмотреть парные коэффициенты корреляции, заменив показатель " подтверждение заболеваемости" на количество смертей к количеству выздоровевших, можно увиеть рост зависимостей. Напрмер между смертностью ко всем случаям заболеваемости и количеством смертей к выздоровевшим, но данные изменения все равно не делают свзь достаточно сильной для дальнейшего изучения. Можно сделать логичный вывод о том, что при увеличении количества смертей по отношению к случаям выздоравления, увеличивается и отношение смертей ко всем случаям заболеваемости.


```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(8, 8))
sns.heatmap(corrmat, square=True)
plt.show()
```

При построении графика необходимо оценить его ясность и доступность изображенной информации. Что касается графика выше можно сказать, что тепловая диаграмма наиболее простой способ оценить силу корреляционных зависимостей. Но при этом на графике выше отсутствует точность в определении самих коэффициентов. Не имея вычисленных коэффициентов корреляции будет сложно достаточно точно их определить по самому графику, что может повлиять на оценку значимости этих коэффициентов или определение мультиколлинеарности.   

```python
plt.hist(data['Deaths / 100 Cases'])
plt.title('Гистограмма распределения ответов по времени')
plt.show()
```

Что касается данного графика, можно отметить, что ему также не хватает точности. Не имея самих данных будет сложно определить точное количество ответов во времени. При этом график демонстрирует общее соотношение времени и ответов, при словестном описании графика на основе точных данных, он может подойти для включения в работу.

```python
nwc_med = data.groupby('WHO Region')['New cases'].median()
nwc_sum = data.groupby('WHO Region')['New cases'].sum()

fig, axs = plt.subplots(2, 1, figsize=(10, 6), sharex=True)
fig.suptitle('Суммарное и медианное число случаев по регионам')
#fig.text('')


axs[0].plot(nwc_sum)
axs[0].set_ylabel('Суммарное число случаев')

axs[1].plot(nwc_med)
axs[1].set_ylabel('Медианное  число случаев')


plt.xlabel('Регионы')

plt.show()
```

Как можно заметить на графике наибольшее число случаев заболеваемости встречается в Америке, при этом наибольшее медианное значение имеет Eastern Mediterranean, что говорит о том, что средние значения в восточном Средиземноморье более высокие. В Европе значения, как медианные, так и обищие достаточно умеренные. А на остальных континентах более низкие.

```python
nwc_med = data.groupby('WHO Region')['Deaths'].median()
nwc_sum = data.groupby('WHO Region')['Deaths'].sum()

fig, axs = plt.subplots(2, 1, figsize=(10, 6), sharex=True)
fig.suptitle('Суммарное и медианное число смертей по регионам')
#fig.text('')


axs[0].plot(nwc_sum)
axs[0].set_ylabel('Суммарное число случаев')

axs[1].plot(nwc_med)
axs[1].set_ylabel('Медианное  число случаев')


plt.xlabel('Регионы')

plt.show()
```

Наибольшее суммарное число смертей наблюдается в Америке, более умеренное - в Европе. При этом наибольшее медианное значение смертности имеет именно Европа, что говорит о более высоких показателях в среднем по регионам. Умеренное медианное значение имеют регионы восточного средиземноморья, что также говорит о достаточно высоких средних показателях, при этом суммарно данные регионы находятся на последнем месте по смертности, что говорит об очень неравномерном распределении числа смертей по регионам ( наличие регионов - выбросов).
