```python
# Создаем объект Figure и задаем размер графика 9x6 дюймов.
plt.figure(figsize = (9, 6))

# Генерируем 10 случайных чисел из стандартного нормального распределения для координат x.
x = np.random.normal(size=10)
# Генерируем 10 случайных чисел из стандартного нормального распределения для координат y.
y = np.random.normal(size=10)

# Строим точечный график (scatter plot).
plt.scatter(x, y, 
            # Задаем цвета каждой точки с помощью списка шестнадцатеричных цветовых кодов.
            # Каждый элемент списка соответствует цвету одной точки. Поскольку у нас 10 точек,
            # в списке 10 цветовых кодов.
            c=[
                '#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd',
                '#8c564b', '#e377c2', '#7f7f7f', '#bcbd22', '#17becf'
            ])

# Добавляем заголовок к графику и устанавливаем размер шрифта.
plt.title('Цвета задали с помощью массива цветовых кодов', fontsize=17)

# Отображаем график.
plt.show()
```

```python
plt.figure(figsize = (11, 8))

plt.title('Различные значения параметров ls и lw', fontsize=17)

plt.plot(x, y,
         linestyle = '-',
         linewidth = 1,
         label="ls = '-',   lw = 1")

plt.plot(x, y + 1,
         linestyle = '--',
         linewidth = 2,
         label="ls = '--',  lw = 2")

plt.plot(x, y + 2,
         linestyle = '-.',
         linewidth = 4, # определите ширину 4
         label="ls = '-.',  lw = 4")

plt.plot(x, y + 3,
         linestyle = ':', # отметьте стиль ':'
         linewidth = 6,
         label="ls = ':',   lw = 6")

plt.legend(title='аргументы ls и lw',
           fontsize=12,
           title_fontsize=13,
           loc='lower right',
           shadow=True)

plt.grid(linestyle='-.', linewidth=2)
plt.show()
```

```python
t = np.linspace(0, 5, 200)* np.ones((7, 200)) + np.random.randn(7, 200).cumsum(axis=1)
a = np.array([0.5, 1, 2.1, -1.3, -0.1, 1.9, -2.9]).reshape(-1, 1)
b = np.array([-3, -4, -10, 4, 1, -5, 6]).reshape(-1, 1)

df = pd.DataFrame((a*t + b).T, columns=list('ABCDEFG'))
```

```python
plt.figure(figsize=(11, 9))
plt.plot(df.values)
plt.legend(labels=df.columns, title='Признаки', title_fontsize=13, fontsize=12)
plt.title('Случайные данные с некоторой корреляцией между признаками', fontsize=15)

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 6))
ax = sns.heatmap(data=df, ax=ax) # отметьте в data матрицу корреляций по df
ax.set_title('Корреляционная матрица', fontsize=15);
```

```python
fig, ax = plt.subplots(figsize=(10, 6))
ax = sns.heatmap(data=df, ax=ax, annot=True)
ax.set_title('Корреляционная матрица с указанием значений', fontsize=15);
```
