## Ассоциативные правила: алгоритмы Apriori и FPgrowth

```python
import pandas as pd
import numpy as np

from mlxtend.frequent_patterns import apriori, fpgrowth, association_rules
```

```python
import warnings
warnings.simplefilter('ignore')
```

```python
df_original = pd.read_excel("emp_data.xls")
df = df_original.copy()
df
```

## Подготовка датасета для использования ассоциативных правил:


**Преобразование бинарных признаков**

```python
for col in '''target
passport
offhour
salary'''.split('\n'):
  df[col] = df[col].astype('bool')

df
```

**Преобразование небинарных признаков**

```python
columns = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
columns
```

```python
for col in columns:
    med_value = df[col].median() 
    df[col] = df[col] > med_value
    df[col] = df[col].astype('bool')
df
```

## Aлгоритм Apriori

```python
apriori_freq_items = apriori(df, min_support = 0.1, use_colnames = True)
apriori_freq_items
```

## Вычисление ассоциативных правил


```python
#support
```

```python
apriori_ar_support = association_rules(apriori_freq_items, metric = 'support', min_threshold = 0.2)
apriori_ar_support
```

```python
apriori_ar_support.sort_values(by='support', ascending=False, inplace=True)
apriori_ar_support
```

```python
#confidence
```

```python
apriori_ar_confidence = association_rules(apriori_freq_items, metric = 'confidence', min_threshold = 0.6)
apriori_ar_confidence.sort_values(by='confidence', ascending=False)
```

```python
#lift
```

```python
apriori_ar_lift = association_rules(apriori_freq_items, metric = 'lift', min_threshold = 1.25)
apriori_ar_lift.sort_values(by='lift', ascending=False)
```

```python
# Чтобы получить негативные правила, возьмем случайные правила и отсортируем по lift
```

```python
apriori_ar_lift_neg = association_rules(apriori_freq_items, metric = 'support', min_threshold = 0.1)
apriori_ar_lift_neg = apriori_ar_lift_neg[apriori_ar_lift_neg['lift'] < 1].sort_values('lift')
apriori_ar_lift_neg.sort_values(by='lift', ascending=True)
```

### Использование FPGrowth

```python
fpgrowth_freq_items = fpgrowth(df, min_support = 0.1, use_colnames = True)
fpgrowth_freq_items
```

```python
#support
```

```python
fpgrowth_ar_support = association_rules(fpgrowth_freq_items, metric = 'support', min_threshold = 0.2)
fpgrowth_ar_support_sorted = fpgrowth_ar_support.sort_values ('support', ascending = False)# Сортировка по Support по убыванию
fpgrowth_ar_support_sorted
```

3 наиболее существенных правила: 

Правило 5: (work_exp) -> (age) : support = 0.456, если у сотрудника большой опыт работы, то с вероятностью 45,6% он в возрасте (более старшего возраста). Что весьма логично, чем больше опыт работы, тем старше должен быть человек, из-за количества отработтанных лет.

Правило 4: (age) -> (work_exp): support = 0.456, если сотрудник более старшего возраста, то с вероятностью 45,6% у него больший опыт работы. Также весьма логично, что чем старше сотрудник, тем больше у него опыта работы.

Правило 39: (interest) -> (coffee): support = 0.267, если у сотрудника более высокий интерес к работе, то с вероятностью 26,7%, он выпивает большее количество чашек кофе на работе. Возможно данныое правило связано с тем, что более заинтересованные работники тратят больше энергии на выполнение задачи и им необходим стимул, например кофе.

Возможно, старшие сотрудники с большим стажем чаще сталкиваются с переработками и, из-за выгорания, планируют увольнение.

Правило 67: (work_exp, boss_educ) -> (age): Confidence = 0.95 Если у сотрудника большой стаж работы и у его руководителя высокое образование, то с вероятностью 95% этот сотрудник старшего возраста. Возможно, руководители с высоким образованием дольше удерживают опытных сотрудников.

Правило 271: (work_exp, offhour) -> (age, target): lift = 2.67 Наличие переработок и больший стаж работы сильно связаны с возрастом сотрудника и планами по увольнению. Это говорит о том, что сотрудники с большим стажем и количеством переработок, более старшего возраста и также соответвенно провели много времени в компании, соответсвенно могут задумываться, например, о пенсии.
