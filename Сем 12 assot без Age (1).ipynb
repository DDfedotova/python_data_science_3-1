## Ассоциативные правила: алгоритмы Apriori и FPgrowth

```python
import pandas as pd
import numpy as np

from mlxtend.frequent_patterns import apriori, fpgrowth, association_rules
```

```python
import warnings
warnings.simplefilter('ignore')
```

```python
df_original = pd.read_excel("emp_data.xls").drop('age', axis=1)
df = df_original.copy()
df
```

## Подготовка датасета для использования ассоциативных правил:


**Преобразование бинарных признаков**

```python
for col in '''target
passport
offhour
salary'''.split('\n'):
  df[col] = df[col].astype('bool')

df
```

**Преобразование небинарных признаков**

```python
columns = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
columns
```

```python
for col in columns:
    med_value = df[col].median() 
    df[col] = df[col] > med_value
    df[col] = df[col].astype('bool')
df
```

## Aлгоритм Apriori

```python
apriori_freq_items = apriori(df, min_support = 0.1, use_colnames = True)
apriori_freq_items
```

## Вычисление ассоциативных правил


```python
#support
```

```python
apriori_ar_support = association_rules(apriori_freq_items, metric = 'support', min_threshold = 0.2)
apriori_ar_support
```

```python
apriori_ar_support.sort_values(by='support', ascending=False, inplace=True)
apriori_ar_support
```

```python
#confidence
```

```python
apriori_ar_confidence = association_rules(apriori_freq_items, metric = 'confidence', min_threshold = 0.6)
apriori_ar_confidence.sort_values(by='confidence', ascending=False)
```

```python
#lift
```

```python
apriori_ar_lift = association_rules(apriori_freq_items, metric = 'lift', min_threshold = 1.25)
apriori_ar_lift.sort_values(by='lift', ascending=False)
```

```python
# Чтобы получить негативные правила, возьмем случайные правила и отсортируем по lift
```

```python
apriori_ar_lift_neg = association_rules(apriori_freq_items, metric = 'support', min_threshold = 0.1)
apriori_ar_lift_neg = apriori_ar_lift_neg[apriori_ar_lift_neg['lift'] < 1].sort_values('lift')
apriori_ar_lift_neg.sort_values(by='lift', ascending=True)
```

### Использование FPGrowth

```python
fpgrowth_freq_items = fpgrowth(df, min_support = 0.1, use_colnames = True)
fpgrowth_freq_items
```

```python
#support
```

```python
fpgrowth_ar_support = association_rules(fpgrowth_freq_items, metric = 'support', min_threshold = 0.2)
fpgrowth_ar_support_sorted = fpgrowth_ar_support.sort_values ('support', ascending = False)# Сортировка по Support по убыванию
fpgrowth_ar_support_sorted
```

3 наиболее существенных правила:

Правило 11: (coffee) -> (interest) : support = 0.267, Сотрудник, который пьет больше кофе на работе с вероятностью 26,7% более заинтересованный. Возможно количество кофе - это показатель, который отражает сколько человеку потребовалось стимуляции, чтобы восстановить силы, так как он сосредоточенно работает, соответсвенно, чем больше, человек употребляет кофе, тем более упорно работает (тратит больше сил), соответсвенно он более заинтересован в своей работе.

Правило 10: (interest) -> (coffee): support = 0.267, Чем более заинтересован сотрудник, тем больше он потребляет кофе ( с вероятностью 26,7%), это также объясняется тем, что сотрудники погруженные в работу, нуждаются в восстановлении сил.

Правило 3: (coffee) -> (work_exp): support = 0.260, Чем больше сотрудники употребляют кофе, тем больше ( с вероятностью 26%) их опыт работы. Возможно это правило отражает, то что сотрудники которые работают дольше, знают что более продуктивная работа подразумевает восстановление сил, соответсвенно они пью больше кофе.

```python
#confidence
```

```python
fpgrowth_ar_confidence = association_rules(fpgrowth_freq_items, metric = 'confidence', min_threshold = 0.6)
fpgrowth_ar_confidence_sorted = fpgrowth_ar_confidence.sort_values ('confidence', ascending = False)# Сортировка пo confidence по убыванию
fpgrowth_ar_confidence_sorted
```

3 наиболее существенных правила:

Правило 13: (work_exp, target) -> (offhour) : support = 0.713, Если сотрудник планирует увольняться и имеет большой стаж работы, то с вероятностью 71,3%, он имеет переработки. Возможно данное правило говорит о том, что сотрудник, который уже хочет уволиться перерабатывает основные часы, например, чтобы закончить определенный проект до своего ухода.

Правило 11: (passport, target) -> (offhour): support = 0.700, Если у сотрудника есть загранпаспорт и он планирует уволиться, то с вероятнотью 70%, он имеет переработки. Возможно сотрудник часто бывающий в командировках (так как имеет загранпаспорт) и также планирует уволиться, имеет выгорание и усталось, от чего не успевает в поставленные часы выполнять все задачи и соотвественно заканчивает работу в дополнительное (личное) время, что приводит к увеличению переработок.

Правило 8: (dist, target) -> (offhour): support = 0.693, Если сотрудник живет далеко от места работы и планирует уволиться, то с вероятностью 69,3%, он имеет переработки. Данное правило отражает, что возможно такой сотрудник устал и ему тяжело работать в данном месте, поэтому он не успевает выполнять работу в рабочие часы или , например, он часто опаздывает из-за расстояния до места работы и ему приходится перерабатывать, чтобы все успеть.


```python
#lift
```

```python
fpgrowth_ar_lift = association_rules(fpgrowth_freq_items, metric = 'lift', min_threshold = 1.25)
fpgrowth_ar_lift_sorted = fpgrowth_ar_lift.sort_values ('lift', ascending = False)# Сортировка пo lift по убыванию
fpgrowth_ar_lift_sorted
```

3 наиболее существенных правила:

Правило 18: (passport, offhour)-> (target): lift = 1.88, Наличие переработок и загран паспорта у сотрудника сильно-положительно связано с его планами по увольнению. Например, сотрудник который часто бывает в командировках ( так как имеет загранпаспорт) и перерабатывает из-за усталости от этих командировок, хочет уволиться в силу очень большой нагрузки, которую тяжело выдержать.

Правило 23: (target) -> (passport, offhour): lift = 1.88 Наличие планов по увольнению у сотрудника сильно связано с наличием у него также загранпаспорта и переработок. Данное правильно объяснить уже тяжелее, чем предыдущее. Если сотрудник планирует уволиться, он перерабатывает чтобы закончить проект до своего ухода и имеет загран паспорт, Например, потому что много летает в командировки или , например, планирует переехать в другую страну после увольнения.

Правило 5: (educ) -> (boss_educ): lift = 1.72 Уровень образования руководителя сильно-положительно зависит от уровня образования сотрудника. Возможно данное правило отражает, что при хорошем коллективе, "плохой" начальник, который не может оптимизировать и улучшить работу сотрудников, долго не задержится на своей высокой должности, то есть ему необходимо иметь высокий уровень образования, чтобы поддерживать уровень всего коллектива.
